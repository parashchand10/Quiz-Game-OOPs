##**Quiz Game - OOPs**

A simple console-based Quiz Game developed using **Python** and **Object-Oriented Programming (OOP)** concepts.

The game presents multiple python programming - related questions, evaluates user input, and calculates scores dynamically based on correct and incorrect answers.

```python
class GuessGame:
    def __init__(self):
        self.score = 0
        self.questions = {"Keyword used to define a function in Python?": "def",
                          "Data type used to store multiple values in sequence?": "list",
                          "Symbol used for comments in Python?": "#",
                          "Loop used to iterate over a sequence?": "for",
                          "Function used to take user input?": "input()"}

    def play(self):
      print("Python Programming Guess Game ?")
      print("-"*20)

      for question, correct_ans in self.questions.items(): # Access keys - Pairs Values
        print(f"Q. {question}")
        answer = input("Enter the answer: ")

        if answer == correct_ans:
            question_score = 1
            self.score = self.score + 1      # Correct answer add by 1
            print("Your answer is Correct")
        else:
            question_score = 0
            print("Your answer is Incorrect")

        print("Score:", question_score)

        print("-"*20)
      print("Your Final Score is:", self.score)

game = GuessGame() # create object
```

```python
game.play()
```
